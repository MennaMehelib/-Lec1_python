# Exercise 1 — Basic class with a method

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2


circle = Circle(5)
print(round(circle.area(), 2))
```

# Exercise 2 — Toggling a boolean attribute

```python
class Light:
    def __init__(self):
        self.is_on = False

    def toggle(self):
        self.is_on = not self.is_on


light = Light()
light.toggle()
light.toggle()
light.toggle()

print(light.is_on)
```

# Exercise 6 — Class using a dictionary internally

```python
class ShoppingCart:
    def __init__(self):
        self.items = {}

    def add_item(self, item, price):
        self.items[item] = price

    def total(self):
        return sum(self.items.values())


cart = ShoppingCart()
cart.add_item("Book", 100)
cart.add_item("Pen", 20)

print(cart.total())
```

# Exercise 9 — A class managing a list of dictionaries

```python
class TodoList:
    def __init__(self):
        self.tasks = []

    def add_task(self, task):
        self.tasks.append({"task": task, "done": False})

    def complete_task(self, task):
        for item in self.tasks:
            if item["task"] == task:
                item["done"] = True
                break


todo = TodoList()
todo.add_task("Study Python")
todo.add_task("Practice OOP")
todo.complete_task("Study Python")

print(todo.tasks)
```

# Exercise 10 — Implementing a Stack

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.items:
            return self.items.pop()
        return None


stack = Stack()
stack.push(1)
stack.push(2)
stack.push(3)

popped = stack.pop()

print("Popped:", popped)
print("Remaining:", stack.items)
```

# Exercise 11 — Custom string representation

```python
class Fraction:
    def __init__(self, num, denom):
        self.num = num
        self.denom = denom

    def value(self):
        return self.num / self.denom

    def __str__(self):
        return f"{self.num}/{self.denom}"


fraction = Fraction(3, 4)

print(fraction)
print(fraction.value())
```

# Exercise 12 — Implementing a Queue

```python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if self.items:
            return self.items.pop(0)
        return None


queue = Queue()
queue.enqueue("a")
queue.enqueue("b")
queue.enqueue("c")

dequeued = queue.dequeue()

print("Dequeued:", dequeued)
print("Remaining:", queue.items)
```

# Exercise 13 — Objects containing other objects

```python
class Movie:
    def __init__(self, title, rating):
        self.title = title
        self.rating = rating


class MovieCollection:
    def __init__(self):
        self.movies = []

    def add(self, movie):
        self.movies.append(movie)

    def average_rating(self):
        if not self.movies:
            return 0
        return sum(movie.rating for movie in self.movies) / len(self.movies)


collection = MovieCollection()
collection.add(Movie("Movie A", 8))
collection.add(Movie("Movie B", 6))
collection.add(Movie("Movie C", 10))

print(collection.average_rating())
```

# Exercise 14 — Method chaining

```python
class Thermostat:
    def __init__(self, temp=20):
        self.temp = temp

    def increase(self, amount):
        self.temp += amount
        return self

    def decrease(self, amount):
        self.temp -= amount
        return self


thermostat = Thermostat()
thermostat.increase(5).increase(3).decrease(2)

print(thermostat.temp)
```

# Exercise 15 — Inventory system with validation

```python
class Inventory2:
    def __init__(self):
        self.stock = {}

    def add_stock(self, item, qty):
        self.stock[item] = self.stock.get(item, 0) + qty

    def remove_stock(self, item, qty):
        if self.stock.get(item, 0) >= qty:
            self.stock[item] -= qty
            return True
        return False


inventory = Inventory2()
inventory.add_stock("pen", 50)
inventory.add_stock("pen", 20)

print("Remove 30:", inventory.remove_stock("pen", 30))
print("Stock:", inventory.stock)

print("Remove 100:", inventory.remove_stock("pen", 100))
print("Stock:", inventory.stock)
```

# Exercise 16 — Class with calculated categories

```python
class BMI:
    def __init__(self, weight, height):
        self.weight = weight
        self.height = height

    def calculate(self):
        return self.weight / (self.height ** 2)

    def category(self):
        bmi = self.calculate()

        if bmi < 18.5:
            return "Underweight"
        elif bmi < 25:
            return "Normal"
        elif bmi < 30:
            return "Overweight"
        else:
            return "Obese"


bmi = BMI(70, 1.75)

print("BMI:", round(bmi.calculate(), 2))
print("Category:", bmi.category())
```

# Exercise 18 — Tracking a running high score

```python
class Game:
    def __init__(self):
        self.score = 0
        self.high_score = 0

    def add_points(self, pts):
        self.score += pts
        if self.score > self.high_score:
            self.high_score = self.score

    def reset(self):
        self.score = 0


game = Game()
game.add_points(50)
game.add_points(30)
game.reset()
game.add_points(20)

print("Final score:", game.score)
print("High score:", game.high_score)
```

# Exercise 19 — Payroll with overtime logic

```python
class Employee3:
    def __init__(self, name, hours, rate):
        self.name = name
        self.hours = hours
        self.rate = rate

    def calculate_pay(self):
        if self.hours <= 40:
            return self.hours * self.rate

        normal_pay = 40 * self.rate
        overtime_hours = self.hours - 40
        overtime_pay = overtime_hours * self.rate * 1.5

        return normal_pay + overtime_pay


employee1 = Employee3("Ahmed", 45, 20)
employee2 = Employee3("Sara", 35, 25)

print(employee1.name, employee1.calculate_pay())
print(employee2.name, employee2.calculate_pay())
```

# Exercise 20 — Comprehensive: a simple linked list

```python
class Node2:
    def __init__(self, value, next=None):
        self.value = value
        self.next = next


class LinkedListSimple:
    def __init__(self):
        self.head = None

    def add(self, value):
        new_node = Node2(value)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current.next is not None:
            current = current.next

        current.next = new_node

    def to_list(self):
        values = []
        current = self.head

        while current is not None:
            values.append(current.value)
            current = current.next

        return values


linked_list = LinkedListSimple()
linked_list.add(10)
linked_list.add(20)
linked_list.add(30)

print(linked_list.to_list())
```
