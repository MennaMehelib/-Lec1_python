```python
import math

def area_of_circle(r):
    return math.pi * r * r

def perimeter_of_circle(r):
    return 2 * math.pi * r

print(area_of_circle(5))
print(perimeter_of_circle(5))
```

```python
def add_all_nums(*args):
    total = 0

    for num in args:
        if not isinstance(num, (int, float)):
            return "All arguments must be numbers"

        total += num

    return total

print(add_all_nums(1, 2, 3, 4))
print(add_all_nums(1, 2, "3"))
```

```python
def convert_celsius_2_fahrenheit(c):
    return (c * 9 / 5) + 32

print(convert_celsius_2_fahrenheit(30))
```

```python
def check_season(month):
    seasons = {
        "December": "Winter",
        "January": "Winter",
        "February": "Winter",

        "March": "Spring",
        "April": "Spring",
        "May": "Spring",

        "June": "Summer",
        "July": "Summer",
        "August": "Summer",

        "September": "Autumn",
        "October": "Autumn",
        "November": "Autumn"
    }

    return seasons.get(month, "Invalid month")

print(check_season("July"))
print(check_season("January"))
```

```python
def calculate_slope(x1, y1, x2, y2):
    return (y2 - y1) / (x2 - x1)

print(calculate_slope(1, 2, 3, 6))
```

```python
import math

def solve_quadratic_eqn(a, b, c):
    D = b ** 2 - 4 * a * c

    if D > 0:
        x1 = (-b + math.sqrt(D)) / (2 * a)
        x2 = (-b - math.sqrt(D)) / (2 * a)
        return x1, x2

    elif D == 0:
        x = -b / (2 * a)
        return x

    else:
        return "No real solution"

print(solve_quadratic_eqn(1, -5, 6))
```

```python
def sum_of_numbers(n):
    total = 0

    for i in range(1, n + 1):
        total += i

    return total


print(sum_of_numbers(5))
print(sum_of_numbers(10))
print(sum_of_numbers(100))
```

```python
def sum_of_odds(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total += i

    return total


print(sum_of_odds(10))
```

```python
def sum_of_even(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 == 0:
            total += i

    return total


print(sum_of_even(10))
```

```python
def reverse_list(items):
    result = []

    for i in range(len(items) - 1, -1, -1):
        result.append(items[i])

    return result


print(reverse_list([1, 2, 3, 4, 5]))
print(reverse_list(["A", "B", "C"]))
```

```python
def evens_and_odds(n):
    evens = 0
    odds = 0

    for i in range(n + 1):
        if i % 2 == 0:
            evens += 1
        else:
            odds += 1

    print(f"The number of odds are {odds}.")
    print(f"The number of evens are {evens}.")


evens_and_odds(100)
```

```python
def calculate_mean(numbers):
    total = 0

    for num in numbers:
        total += num

    return total / len(numbers)


print(calculate_mean([1, 2, 3, 4, 5]))
```

```python
def calculate_median(numbers):
    numbers = sorted(numbers)

    n = len(numbers)
    middle = n // 2

    if n % 2 != 0:
        return numbers[middle]
    else:
        return (numbers[middle - 1] + numbers[middle]) / 2


print(calculate_median([5, 2, 1, 4, 3]))
print(calculate_median([1, 2, 3, 4]))
```

```python
def calculate_mode(numbers):
    counts = {}

    for num in numbers:
        if num in counts:
            counts[num] += 1
        else:
            counts[num] = 1

    max_count = max(counts.values())

    for num in counts:
        if counts[num] == max_count:
            return num


print(calculate_mode([1, 2, 2, 3, 4]))
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


print(is_prime(7))
print(is_prime(10))
```

```python
def all_unique(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return False

    return True


print(all_unique([1, 2, 3, 4]))
print(all_unique([1, 2, 2, 4]))
```

```python
def same_data_type(items):
    if len(items) == 0:
        return True

    first_type = type(items[0])

    for item in items:
        if type(item) != first_type:
            return False

    return True


print(same_data_type([1, 2, 3, 4]))
print(same_data_type([1, "2", 3]))
print(same_data_type(["A", "B", "C"]))
```

```python
import keyword

def is_valid_variable(variable):
    if not variable.isidentifier():
        return False

    if keyword.iskeyword(variable):
        return False

    return True


print(is_valid_variable("student_name"))
print(is_valid_variable("2student"))
print(is_valid_variable("student-name"))
print(is_valid_variable("class"))
```
